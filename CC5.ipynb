```python
import numpy as np
```

```python
#Ques 1
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python
#Ques 2
arr = np.array([1, 2, 3, 6, 4, 5])
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]
print("Most frequent value:", most_frequent)
print("Frequency:", counts[np.argmax(counts)])
print("Indices:", indices)
```

```python
#Ques 3
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
#Ques 5
import numpy as np
Arnav = np.linspace(10, 100, 25)
print("Array:")
print(Arnav)
print("Dimensions:", Arnav.ndim)
print("Shape:", Arnav.shape)
print("Total elements:", Arnav.size)
print("Data type:", Arnav.dtype)
print("Total bytes:", Arnav.nbytes)
transpose_Arnav = Arnav.reshape(1, 25)

print("Transpose using reshape():")
print(transpose_Arnav)


print("Transpose using T:")
print(Arnav.T)
```

```python
#Ques 6
ucs420_Arnav = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print("Original Array:")
print(ucs420_Arnav)

print("Mean:", np.mean(ucs420_Arnav))

print("Median:", np.median(ucs420_Arnav))

print("Maximum:", np.max(ucs420_Arnav))

print("Minimum:", np.min(ucs420_Arnav))

print("Unique Elements:", np.unique(ucs420_Arnav))

reshaped_ucs420_Arnav = ucs420_Arnav.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Arnav)

resized_ucs420_Arnav = np.resize(ucs420_Arnav, (2, 3))

print("Resized Array (2 x 3):")
print(resized_ucs420_Arnav)
```
