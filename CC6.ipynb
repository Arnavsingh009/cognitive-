```python
import numpy as np
```

```python
#Ques 1
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Incorrect tempratures:",temperature)
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)
```

```python
fahrenheit = (9/5) * corrected_temperature + 32
print("Temperature in Fahrenheit:", fahrenheit)
```

```python
readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings_greater_32)
```

```python
count = np.sum(corrected_temperature > 32)
print("Number of readings greater than 32°C:", count)
```

```python
#Ques 2
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

```python
total_steps = np.sum(steps)
print("Total steps:", total_steps)
```

```python
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
```

```python
maximum = np.max(steps)
minimum = np.min(steps)
print("Maximum steps:", maximum)
print("Minimum steps:", minimum)
```

```python
day_total = np.sum(steps, axis=0)
print("Total steps for each day:", day_total)
```

```python
user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)
```

```python
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", position)
print("Maximum steps:", steps[position])
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
```

```python
ravel_array = matrix.ravel()
print("Using ravel():", ravel_array)
```

```python
#Ques 4
y = np.array([40, 65, 30, 85])
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
print("Transpose of X:")
print(X.T)
```

```python
XT_X = X.T @ X

print("X.T @ X:")
print(XT_X)
```

```python
inverse = np.linalg.inv(X.T @ X)

print("Inverse of X.T @ X:")
print(inverse)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Beta coefficients:")
print(beta)
```

```python
new_user = np.array([5, 40, 7])
new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("Predicted assistance score:", prediction)
```
