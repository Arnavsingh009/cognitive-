```python
#Assignment 5.3
n=int(input("What is the number "))
sum=0
for i in range(2,n+1):
   prime=True
   for j in range(2,int(i**0.5)+1):
    if(i%j==0):
      prime=False
      break
   if prime:
    sum=sum+i

print(sum)
```
